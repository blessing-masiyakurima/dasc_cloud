```python
import pandas as pd
import streamlit as st
import plotly.express as px

st.set_page_config(
    page_title="ChartPulse",
    page_icon="🎵",
    layout="wide"
)

st.title("🎵 ChartPulse")
st.subheader("Spotify Album Trends Explorer")

@st.cache_data
def load_data():
    return pd.read_csv(
        "charts_albums_weekly.csv",
        low_memory=False
    )

df = load_data()

# Sidebar
countries = sorted(df["country"].unique())

selected_country = st.sidebar.selectbox(
    "Country",
    countries
)

filtered_df = df[
    df["country"] == selected_country
].copy()

# KPI Cards
col1, col2, col3 = st.columns(3)

col1.metric(
    "Albums",
    filtered_df["album_name"].nunique()
)

col2.metric(
    "Artists",
    filtered_df["artist_names"].nunique()
)

col3.metric(
    "Rows",
    f"{len(filtered_df):,}"
)

st.divider()

# Longest Charting Albums
st.header("🏆 Longest Charting Albums")

longevity = (
    filtered_df.groupby(
        ["album_name", "artist_names"]
    )["weeks_on_chart"]
    .max()
    .reset_index()
    .sort_values(
        "weeks_on_chart",
        ascending=False
    )
    .head(20)
)

fig = px.bar(
    longevity,
    x="weeks_on_chart",
    y="album_name",
    color="weeks_on_chart",
    orientation="h",
    hover_data=["artist_names"],
    title="Top 20 Albums by Weeks on Chart"
)

fig.update_layout(
    yaxis={
        "categoryorder": "total ascending"
    }
)

st.plotly_chart(
    fig,
    use_container_width=True
)
```
